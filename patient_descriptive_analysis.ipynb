```python
#!pip install tableone
import pandas as pd
from tableone import TableOne
import warnings

#import ace_tools as tools
warnings.filterwarnings("ignore")
pd.set_option('display.max_rows', 200)  # Set to 200 or more, higher than 105
pd.set_option('display.max_columns', None)  # Show all columns if needed
```

```python
patient_level_df = pd.read_csv('alert_analysis/data/main_data_2022/df_patients_level_data.csv')
print(patient_level_df.shape)
patient_level_df.head(10)
```

```python
# Define categorical and continuous variables
categorical_vars = [
    "age_category",
    'unit_category_ud',  # Added: unit category
    'unit_Internal',
    'unit_Internal-Covid19',
    'unit_Surgery',
    'unit_Gynecology',
    'unit_Cardiology',
    'unit_Emergency',
    'unit_Geriatric',
    'unit_Hematology',
    'unit_Nephrology',
    'unit_Oncology',
    'gender',
    'chronic_med_ud',  # Added: chronic medication
    # Disease columns - Added
    'kidney_disease',
    'hepatic_disease',
    'diabetes_disease',
    'ischemic_heart_disease',
    'copd_disease',
    'cerebrovascular_disease',
    'peptic_ulcer_disease',
    'dementia_disease',
    'oncological_disease',
    'hemato_oncological_disease',
    'hypertension_disease',
    'atrial_fibrillation_disease',
    'hyperlipidemia_disease',
    'congestive_heart_failure_disease',
    'obesity_disease',
]

continuous_vars = [
    "age",
    'survival_rate_10y_age_adj_mean',
    'charlson_score_age_adj_mean',
    'hospital_days_mean',
    'chronic_med_count_mean',
    'diagnosis_count_mean',
]

#non_normal = ["age"]

# Convert categorical columns to strings to avoid type comparison errors
# This ensures all categorical values are strings before TableOne processes them
for col in categorical_vars:
    if col in patient_level_df.columns:
        patient_level_df[col] = patient_level_df[col].astype(str)

# Generate summary table
table = TableOne(
    patient_level_df,
    categorical=categorical_vars,
    groupby="gender",
    continuous=continuous_vars,
    #nonnormal=non_normal,
    pval=True,  # Add p-values
    missing=True  # Show missing values
)

# Print the table
print(table.tabulate(tablefmt="pipe"))
```

```python
# Comprehensive TableOne with ALL continuous and categorical variables
# Overall summary (no grouping) - includes all variables

# Ensure all categorical columns are strings
for col in categorical_vars:
    if col in patient_level_df.columns:
        patient_level_df[col] = patient_level_df[col].astype(str)

# Binary variables that should show only "1" category
binary_vars = [col for col in categorical_vars if col.startswith('unit_') and col != 'unit_category_ud'] + \
              [col for col in categorical_vars if col.endswith('_disease')]

# Create comprehensive table with all variables
table_all = TableOne(
    patient_level_df,
    categorical=categorical_vars,
    continuous=continuous_vars,
    pval=False,  # No p-values needed for overall summary
    missing=False,  # Hide missing column
    overall=True  # Show overall column
)

# Filter the TableOne output: for binary variables, keep only category "1"
table_df = table_all.tableone

if table_df is not None and not table_df.empty and isinstance(table_df.index, pd.MultiIndex):
    # Build list of indices to keep
    keep_indices = []
    
    for idx in table_df.index:
        var_name_full = str(idx[0])  # e.g., 'unit_Internal, n (%)'
        category = str(idx[1]) if len(idx) > 1 else ""
        
        # Extract base variable name
        var_name_base = var_name_full.split(',')[0].strip()
        
        # Check if binary variable
        is_binary = var_name_base in binary_vars
        
        if not is_binary:
            # Keep all non-binary rows
            keep_indices.append(idx)
        elif is_binary and category.strip() == "1":
            # Keep only "1" category for binary variables
            keep_indices.append(idx)
        # Skip binary variables with category "0" or other values
    
    # Filter DataFrame
    table_filtered = table_df.loc[keep_indices]
    
    # Format and print
    print("=" * 80)
    print("COMPREHENSIVE DESCRIPTIVE STATISTICS - ALL VARIABLES")
    print("=" * 80)
    
    # Format output to match TableOne style
    from tabulate import tabulate
    
    output_rows = []
    for idx in table_filtered.index:
        var_name_full = str(idx[0])  # e.g., 'unit_Internal, n (%)'
        category = str(idx[1]) if len(idx) > 1 else ""
        overall_val = str(table_filtered.loc[idx, 'Overall'])
        
        if category == "":
            # Continuous variable or special row (like 'n')
            output_rows.append([var_name_full, "", overall_val])
        else:
            # Categorical variable with category
            # Show variable name on first row, category on second row (TableOne style)
            var_name_base = var_name_full.split(',')[0].strip()
            if var_name_base in binary_vars:
                # For binary variables, show: variable name row, then category "1" row
                output_rows.append([var_name_full, "", ""])  # Variable header
                output_rows.append(["", category, overall_val])  # Category "1" with value
            else:
                # Non-binary categorical (shouldn't be filtered, but handle)
                output_rows.append([var_name_full, "", ""])
                output_rows.append(["", category, overall_val])
    
    print(tabulate(output_rows, headers=["", "", "Overall"], tablefmt="pipe", showindex=False))
else:
    # Fallback
    print("=" * 80)
    print("COMPREHENSIVE DESCRIPTIVE STATISTICS - ALL VARIABLES")
    print("=" * 80)
    print(table_all.tabulate(tablefmt="pipe"))
```

```python
# Comprehensive TableOne with ALL variables grouped by gender
# Includes all continuous and categorical variables with gender comparison

# Ensure all categorical columns are strings
for col in categorical_vars:
    if col in patient_level_df.columns:
        patient_level_df[col] = patient_level_df[col].astype(str)

# Create comprehensive table grouped by gender with all variables
table_all_by_gender = TableOne(
    patient_level_df,
    categorical=categorical_vars,
    continuous=continuous_vars,
    groupby="gender",
    pval=True,  # Add p-values for gender comparison
    missing=False,  # Don't Show missing values
    overall=True  # Show overall column
)

# Print the comprehensive table
print("=" * 80)
print("COMPREHENSIVE DESCRIPTIVE STATISTICS - ALL VARIABLES BY GENDER")
print("=" * 80)
print(table_all_by_gender.tabulate(tablefmt="pipe"))

# Save to Excel and CSV
import os
from datetime import datetime

# Create output directory if it doesn't exist
output_dir = "output"
os.makedirs(output_dir, exist_ok=True)

# Get the underlying DataFrame
table_df = table_all_by_gender.tableone

# Save to CSV
csv_filename = os.path.join(output_dir, "comprehensive_descriptive_statistics_by_gender.csv")
table_df.to_csv(csv_filename, index=True)
print(f"\n✓ Table saved to CSV: {csv_filename}")

# Save to Excel
try:
    excel_filename = os.path.join(output_dir, "comprehensive_descriptive_statistics_by_gender.xlsx")
    table_df.to_excel(excel_filename, index=True, engine='openpyxl')
    print(f"✓ Table saved to Excel: {excel_filename}")
except ImportError:
    print("\n⚠ Warning: openpyxl not installed. Install with: pip install openpyxl")
    print("  CSV file saved successfully.")
except Exception as e:
    print(f"\n⚠ Warning: Could not save Excel file: {str(e)}")
    print("  CSV file saved successfully.") 
```
